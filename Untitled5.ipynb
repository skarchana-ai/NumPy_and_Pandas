NumPy Array Operations

```python
import numpy as np
a= np.array([10,20,30])
b= np.array([1,2,3])
```

```python
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a**2)
```

Mathematical Functions

```python
arr=np.array([1,4,9])
```

```python
print(np.sqrt(arr))
```

```python
print(np.exp(arr))
```

```python
print(np.log(arr))
```

Statistics

```python
arr=np.array([10,20,30,40])
```

```python
print(arr.sum())
```

```python
print(arr.mean())
```

```python
print(arr.min())
```

```python
print(arr.max())
```

```python
print(np.median(arr))
```

```python
print(arr.std())
```

Axis

```python
arr=np.array([
    [1,2,3],
    [4,5,6]
])
```

```python
print(arr.sum(axis=0))
```

```python
print(arr.sum(axis=1))
```
