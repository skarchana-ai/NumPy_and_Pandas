Pandas

```python
import pandas as pd
```

A Series is a one dimensional labelled data structure.

```python
marks=pd.Series([80,75,90,85])
```

```python
print(marks)
```

80,75,90,85 --> values
0,1,2,3 -->labels/index
It is one-dimensional --> like a single column of data

We can also create our own labels custom Index

```python
marks=pd.Series(
    [80,75,90],
    index=["john",
           "anu",
           "Rahul"
           ]
)
```

```python
marks
```

loc and iloc are used to select data from a Series or DataFrame.

loc --> selects using labels

iloc--> selects using integer positions

```python
df.loc[0]
```

```python
df.loc[
    0,
    "Name"
]
```

```python
df.loc[
    0:2,
    ["Name","Marks"]
]
```

```python
df.iloc[0]
```

```python
df.iloc[
    0:2,
    0:2
]
```

Position: 0,1,2,3

Label: John Anu Rahul Sara

Marks: 80 75 90 85
